```python
# Simple RAG with LlamaIndex and Google Gemini
from llama_index.core import VectorStoreIndex, SimpleDirectoryReader
from llama_index.readers.web import BeautifulSoupWebReader
from llama_index.embeddings.google_genai import GoogleGenAIEmbedding
from llama_index.llms.google_genai import GoogleGenAI
import os

# Set your API key
api_key = "AIzaSyBWC5p7v_84GCLJ7lZOyEX7LwN2KJBF9zc"

# Set up LLM and embeddings
llm = GoogleGenAI(
    model="gemini-2.5-flash-lite",
    api_key = api_key
)
embed_model = GoogleGenAIEmbedding(model_name="text-embedding-004",
api_key=api_key)

documents = BeautifulSoupWebReader().load_data(
    ["https://souvikmajumder31.medium.com/langchain-interview-questions-answers-a72ea8aeddd9"]
)

# Create index
index = VectorStoreIndex.from_documents(documents, embed_model=embed_model)

# Create query engine
query_engine = index.as_query_engine(llm=llm)

# Query the system
query = "Give 7th Common Interview Question on langchain and its answer"
response = query_engine.query(query)

print("Answer:", response)
```
