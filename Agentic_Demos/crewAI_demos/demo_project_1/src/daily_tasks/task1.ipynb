```python
import re
from datetime import datetime
```

```python
log_file = "/workspace/Agentic_AI_Demos_02/Agentic_Demos/crewAI_demos/demo_project_1/logs.txt"
```

```python
target_time = "2025-09-01 11:52:01"
```

```python
with open(log_file, "r", encoding="utf-8") as f:
    log_lines = f.readlines()

print(f"Total log lines loaded: {len(log_lines)}")
```

```python
entries = []
current_entry = []

for line in log_lines:
    if re.match(r"^\d{4}-\d{2}-\d{2} \d{2}:\d{2}:\d{2}:", line):
        if current_entry:  # save the previous entry
            entries.append("".join(current_entry).strip())
        current_entry = [line]  # start new entry
    else:
        current_entry.append(line)  # continuation of current log

# Add the last entry
if current_entry:
    entries.append("".join(current_entry).strip())
```

```python
matching_entries = [entry for entry in entries if target_time in entry]

if matching_entries:
    for e in matching_entries:
        print("="*80)
        print(e)
        print("="*80)
else:
    print(f"No logs found for {target_time}")
```
